```python
#2
import pandas as pd
import matplotlib as mt
import tqdm as tq
import numpy as np
import matplotlib.pyplot as plt
```

```python
#1
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
```

```python
#3
distribuicoes["Normal"]["dados"] = np.random.normal(0, 1, 1000)
distribuicoes["Exponencial"]["dados"] = np.random.exponential(2, 1000)
distribuicoes["Uniforme"]["dados"] = np.random.uniform(-1, 1, 1000)
```

```python
#4
dados_normal = distribuicoes["Normal"]["dados"]
distribuicoes["Normal"]["estatisticas"] = {
    "Média": np.mean(dados_normal),
    "Mediana": np.median(dados_normal),
    "Variância": np.var(dados_normal),
    "Desvio padrão": np.std(dados_normal)
}

dados_exp = distribuicoes["Exponencial"]["dados"]
distribuicoes["Exponencial"]["estatisticas"] = {
    "Média": np.mean(dados_exp),
    "Mediana": np.median(dados_exp),
    "Variância": np.var(dados_exp),
    "Desvio padrão": np.std(dados_exp)
}

dados_uni = distribuicoes["Uniforme"]["dados"]
distribuicoes["Uniforme"]["estatisticas"] = {
    "Média": np.mean(dados_uni),
    "Mediana": np.median(dados_uni),
    "Variância": np.var(dados_uni),
    "Desvio padrão": np.std(dados_uni)
}
```

```python
#5
plt.figure(figsize=(6, 4)) 
plt.hist(distribuicoes["Normal"]["dados"], bins=30, color='blue', edgecolor='black')
plt.title("Distribuição Normal")
plt.show()

plt.figure(figsize=(6, 4)) 
plt.hist(distribuicoes["Exponencial"]["dados"], bins=30, color='green', edgecolor='black')
plt.title("Distribuição Exponencial")
plt.show()

plt.figure(figsize=(6, 4))
plt.hist(distribuicoes["Uniforme"]["dados"], bins=30, color='red', alpha=0.7, edgecolor='black')
plt.title("Distribuição Uniforme")
plt.show()
```

```python
#6
tabela = {}


estatisticas_normal = distribuicoes["Normal"]["estatisticas"]
tabela["Normal"] = estatisticas_normal


estatisticas_exponencial = distribuicoes["Exponencial"]["estatisticas"]
tabela["Exponencial"] = estatisticas_exponencial


estatisticas_uniforme = distribuicoes["Uniforme"]["estatisticas"]
tabela["Uniforme"] = estatisticas_uniforme


df_stats = pd.DataFrame(tabela)


df_final = df_stats.T

print(df_final)
```
