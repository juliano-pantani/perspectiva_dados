```python
#2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#1
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None},
    "Poisson": {"params": {"lambda": 1}, "dados": None, "estatisticas": None},
    "Binomial": {"params": {"n": 10, "p": 0.5}, "dados": None, "estatisticas": None}
}
```

```python
#3
distribuicoes["Normal"]["dados"] = np.random.normal(0, 1, 1000)
distribuicoes["Exponencial"]["dados"] = np.random.exponential(2, 1000)
distribuicoes["Uniforme"]["dados"] = np.random.uniform(-1, 1, 1000)
```

```python
#4
dados_normal = distribuicoes["Normal"]["dados"]
distribuicoes["Normal"]["estatisticas"] = {
    "Média": np.mean(dados_normal),
    "Mediana": np.median(dados_normal),
    "Variância": np.var(dados_normal),
    "Desvio padrão": np.std(dados_normal)
}

dados_exp = distribuicoes["Exponencial"]["dados"]
distribuicoes["Exponencial"]["estatisticas"] = {
    "Média": np.mean(dados_exp),
    "Mediana": np.median(dados_exp),
    "Variância": np.var(dados_exp),
    "Desvio padrão": np.std(dados_exp)
}

dados_uni = distribuicoes["Uniforme"]["dados"]
distribuicoes["Uniforme"]["estatisticas"] = {
    "Média": np.mean(dados_uni),
    "Mediana": np.median(dados_uni),
    "Variância": np.var(dados_uni),
    "Desvio padrão": np.std(dados_uni)
}
```

```python
#5
plt.figure(figsize=(6, 4)) 
plt.hist(distribuicoes["Normal"]["dados"], bins=30, color='blue', edgecolor='black')
plt.title("Distribuição Normal")
plt.show()

plt.figure(figsize=(6, 4)) 
plt.hist(distribuicoes["Exponencial"]["dados"], bins=30, color='green', edgecolor='black')
plt.title("Distribuição Exponencial")
plt.show()

plt.figure(figsize=(6, 4))
plt.hist(distribuicoes["Uniforme"]["dados"], bins=30, color='red', alpha=0.7, edgecolor='black')
plt.title("Distribuição Uniforme")
plt.show()
```

```python
#6
tabela = {}


estatisticas_normal = distribuicoes["Normal"]["estatisticas"]
tabela["Normal"] = estatisticas_normal


estatisticas_exponencial = distribuicoes["Exponencial"]["estatisticas"]
tabela["Exponencial"] = estatisticas_exponencial


estatisticas_uniforme = distribuicoes["Uniforme"]["estatisticas"]
tabela["Uniforme"] = estatisticas_uniforme


df_stats = pd.DataFrame(tabela)

print(df_stats)
```

```python
#7
distribuicoes["Poisson"]["dados"] = np.random.poisson( 1, 1000)
distribuicoes["Binomial"]["dados"] = np.random.binomial(10,.5,1000)
```

```python
#8
dados_normal = distribuicoes["Normal"]["dados"]
distribuicoes["Normal"]["estatisticas"] = {
    "Média": np.mean(dados_normal),
    "Mediana": np.median(dados_normal),
    "Variância": np.var(dados_normal),
    "Desvio padrão": np.std(dados_normal),
    "Mínimo": np.min(dados_normal),
    "Máximo": np.max(dados_normal),
    "Quartil 1": np.percentile(dados_normal,.25),
    "Quartil 3": np.percentile(dados_normal,.75)
}

dados_exp = distribuicoes["Exponencial"]["dados"]
distribuicoes["Exponencial"]["estatisticas"] = {
    "Média": np.mean(dados_exp),
    "Mediana": np.median(dados_exp),
    "Variância": np.var(dados_exp),
    "Desvio padrão": np.std(dados_exp),
    "Mínimo": np.min(dados_exp),
    "Máximo": np.max(dados_exp),
    "Quartil 1": np.percentile(dados_exp,.25),
    "Quartil 3": np.percentile(dados_exp,.75)
}

dados_uni = distribuicoes["Uniforme"]["dados"]
distribuicoes["Uniforme"]["estatisticas"] = {
    "Média": np.mean(dados_uni),
    "Mediana": np.median(dados_uni),
    "Variância": np.var(dados_uni),
    "Desvio padrão": np.std(dados_uni),
    "Mínimo": np.min(dados_uni),
    "Máximo": np.max(dados_uni),
    "Quartil 1": np.percentile(dados_uni,.25),
    "Quartil 3": np.percentile(dados_uni,.75)
}

dados_poisson = distribuicoes["Poisson"]["dados"]
distribuicoes["Poisson"]["estatisticas"] = {
    "Média": np.mean(dados_poisson),
    "Mediana": np.median(dados_poisson),
    "Variância": np.var(dados_poisson),
    "Desvio padrão": np.std(dados_poisson),
    "Mínimo": np.min(dados_poisson),
    "Máximo": np.max(dados_poisson),
    "Quartil 1": np.percentile(dados_poisson,.25),
    "Quartil 3": np.percentile(dados_poisson,.75)
}

dados_binomial = distribuicoes["Binomial"]["dados"]
distribuicoes["Binomial"]["estatisticas"] = {
    "Média": np.mean(dados_binomial),
    "Mediana": np.median(dados_binomial),
    "Variância": np.var(dados_binomial),
    "Desvio padrão": np.std(dados_binomial),
    "Mínimo": np.min(dados_binomial),
    "Máximo": np.max(dados_binomial),
    "Quartil 1": np.percentile(dados_binomial,.25),
    "Quartil 3": np.percentile(dados_binomial,.75)
}


estatisticas_normal = distribuicoes["Normal"]["estatisticas"]
tabela["Normal"] = estatisticas_normal

estatisticas_exponencial = distribuicoes["Exponencial"]["estatisticas"]
tabela["Exponencial"] = estatisticas_exponencial

estatisticas_uniforme = distribuicoes["Uniforme"]["estatisticas"]
tabela["Uniforme"] = estatisticas_uniforme

estatisticas_poisson = distribuicoes["Poisson"]["estatisticas"]
tabela["Poisson"] = estatisticas_poisson

estatisticas_binomial = distribuicoes["Binomial"]["estatisticas"]
tabela["Binomial"] = estatisticas_binomial


df_stats = pd.DataFrame(tabela)

df_final = df_stats.T

print(df_final)
```

```python
#9
plt.figure(figsize=(10, 6))

dados_boxplot = [
    distribuicoes["Normal"]["dados"],
    distribuicoes["Exponencial"]["dados"],
    distribuicoes["Uniforme"]["dados"],
    distribuicoes["Poisson"]["dados"],
    distribuicoes["Binomial"]["dados"]
]

labels = ["Normal", "Exponencial", "Uniforme", "Poisson", "Binomial"]

plt.boxplot(dados_boxplot, labels=labels)
plt.title("Boxplot de Todas as Distribuições", fontsize=14, fontweight='bold')
plt.ylabel("Valores", fontsize=12)
plt.xlabel("Distribuições", fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
