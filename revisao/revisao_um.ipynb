```python
#1
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
#2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#3
distribuicoes["Normal"]["dados"] = np.random.normal(0,1,1000)
distribuicoes["Exponencial"]["dados"] = np.random.exponential(2,1000)
distribuicoes["Uniforme"]["dados"] = np.random.uniform(-1,1,1000)
#4
a = ["Normal","Exponencial","Uniforme"]
for i in a:
    distribuicoes[i]["estatisticas"] = {"media" : np.mean(distribuicoes[i]["dados"]),
                                        "mediana": np.median(distribuicoes[i]["dados"]),
                                        "variancia": np.var(distribuicoes[i]["dados"]),
                                        "desvio padrao": np.std(distribuicoes[i]["dados"])}
distribuicoes
```

```python

#5
for i in a:
    plt.hist(distribuicoes[i]["dados"],bins = 15)
    plt.title("Histograma " + i)
    plt.show()
#6
a = ["Normal","Exponencial","Uniforme"]
df = {}
for i in a:
    df[i] = {"media" : np.mean(distribuicoes[i]["dados"]),
             "mediana": np.median(distribuicoes[i]["dados"]),
             "variancia": np.var(distribuicoes[i]["dados"]),
             "desvio padrao": np.std(distribuicoes[i]["dados"])}
df = pd.DataFrame(df)
df
```

```python
#7
distribuicoes["Poisson"] = {"params": {"scale": 1}, "dados": None, "estatisticas": None}
distribuicoes["Poisson"]["dados"] = np.random.poisson(1,1000)
a = ["Normal","Exponencial","Uniforme","Poisson"]
c = "Poisson"
df[c] = {"media" : np.mean(distribuicoes[c]["dados"]),
             "variancia": np.median(distribuicoes[c]["dados"]),
             "variancia": np.var(distribuicoes[c]["dados"]),
             "desvio padrao": np.std(distribuicoes[c]["dados"])}
#8
df2 = {}
for i in a:
    df2[i] = {"media" : np.mean(distribuicoes[i]["dados"]),
             "variancia": np.median(distribuicoes[i]["dados"]),
             "variancia": np.var(distribuicoes[i]["dados"]),
             "desvio padrao": np.std(distribuicoes[i]["dados"]),
             "min" : np.min(distribuicoes[i]["dados"]),
             "max": np.max(distribuicoes[i]["dados"]),
             "quartil1": np.quantile(distribuicoes[i]["dados"],0.25),
             "quartil3": np.quantile(distribuicoes[i]["dados"],0.75)}
df2 = pd.DataFrame(df2)
df2
```

```python
#9
fig, axs = plt.subplots(1, 4, figsize=(15, 5))
axs[0].boxplot(distribuicoes["Normal"]["dados"])
axs[0].set_title('Normal')
axs[0].set_xlabel(' ')
axs[0].set_ylabel('Valor')
axs[1].boxplot(distribuicoes["Exponencial"]["dados"])
axs[1].set_title('Exponencial')
axs[1].set_xlabel(' ')
axs[1].set_ylabel('Valor')
axs[2].boxplot(distribuicoes["Uniforme"]["dados"])
axs[2].set_title('Uniforme')
axs[2].set_xlabel(' ')
axs[2].set_ylabel('Valor')
axs[3].boxplot(distribuicoes["Poisson"]["dados"])
axs[3].set_title('Poisson')
axs[3].set_xlabel(' ')
axs[3].set_ylabel('Valor')
plt.tight_layout()
plt.show()
```
