```python
#primeiro passo, criando o dicionário principal
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
```

```python
#segundo passo, importar bibliotecas
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#terceiro passo, gerar os dados
distribuicoes["Normal"]["dados"] = np.random.normal(0, 1, size=1000)
distribuicoes["Exponencial"]["dados"] = np.random.exponential(2, size=1000)
distribuicoes["Uniforme"]["dados"] = np.random.uniform(-1, 1, size=1000)
```

```python
#passo quatro, calcular estatísticas
distribuicoes["Normal"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Normal"]["dados"]),
    "mediana": np.median(distribuicoes["Normal"]["dados"]),
    "variancia": np.var(distribuicoes["Normal"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Normal"]["dados"])
}

distribuicoes["Exponencial"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Exponencial"]["dados"]),
    "mediana": np.median(distribuicoes["Exponencial"]["dados"]),
    "variancia": np.var(distribuicoes["Exponencial"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Exponencial"]["dados"])
}

distribuicoes["Uniforme"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Uniforme"]["dados"]),
    "mediana": np.median(distribuicoes["Uniforme"]["dados"]),
    "variancia": np.var(distribuicoes["Uniforme"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Uniforme"]["dados"])
}
```

```python
#passo cinco, criar histograma
plt.title('Hist. Normal')
plt.hist(distribuicoes["Normal"]["dados"], 10, rwidth=0.9)
plt.show()
plt.title('Hist. Exponencial')
plt.hist(distribuicoes["Exponencial"]["dados"], 10, rwidth=0.9)
plt.show()
plt.title('Hist. Uniforme')
plt.hist(distribuicoes["Uniforme"]["dados"], 10, rwidth=0.9)
plt.show()
```

```python
#passo seis, organizar dados com pandas
data = {
  "Normal": distribuicoes["Normal"]["estatisticas"],
  "Exponencial": distribuicoes["Exponencial"]["estatisticas"],
  "Uniforme": distribuicoes["Uniforme"]["estatisticas"]
}
data_frame = pd.DataFrame(data)
print(data_frame)
data_frame.T
```

```python
#passo sete
#sete.um gerando dados para poisson
distribuicoes["Poisson"] = {"params": {"lam": 3}, "dados": None, "estatisticas": None}
#sete.dois gerando os dados
distribuicoes["Poisson"]["dados"] = np.random.poisson(3, size=1000)
#sete.tres calculando as estatísticas  
distribuicoes["Poisson"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Poisson"]["dados"]),
    "mediana": np.median(distribuicoes["Poisson"]["dados"]),
    "variancia": np.var(distribuicoes["Poisson"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Poisson"]["dados"])
}
#sete.quatro adicionando novos dados
data1 = {
  "Normal": distribuicoes["Normal"]["estatisticas"],
  "Exponencial": distribuicoes["Exponencial"]["estatisticas"],
  "Uniforme": distribuicoes["Uniforme"]["estatisticas"],
  "Poisson": distribuicoes["Poisson"]["estatisticas"]
}
data_frame1 = pd.DataFrame(data1)
print(data_frame1)
data_frame1.T
```

```python
#passo oito, calculando novas estatísticas

distribuicoes["Normal"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Normal"]["dados"]),
    "mediana": np.median(distribuicoes["Normal"]["dados"]),
    "variancia": np.var(distribuicoes["Normal"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Normal"]["dados"]),
    "min": np.min(distribuicoes["Normal"]["dados"]),
    "max": np.max(distribuicoes["Normal"]["dados"]),
    "primeiro_quartil": np.percentile(distribuicoes["Normal"]["dados"], 25),
    "terceiro_quartil": np.percentile(distribuicoes["Normal"]["dados"], 75)
}

distribuicoes["Exponencial"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Exponencial"]["dados"]),
    "mediana": np.median(distribuicoes["Exponencial"]["dados"]),
    "variancia": np.var(distribuicoes["Exponencial"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Exponencial"]["dados"]),
    "min": np.min(distribuicoes["Exponencial"]["dados"]),
    "max": np.max(distribuicoes["Exponencial"]["dados"]),
    "primeiro_quartil": np.percentile(distribuicoes["Exponencial"]["dados"], 25),
    "terceiro_quartil": np.percentile(distribuicoes["Exponencial"]["dados"], 75)
}

distribuicoes["Uniforme"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Uniforme"]["dados"]),
    "mediana": np.median(distribuicoes["Uniforme"]["dados"]),
    "variancia": np.var(distribuicoes["Uniforme"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Uniforme"]["dados"]),
    "min": np.min(distribuicoes["Uniforme"]["dados"]),
    "max": np.max(distribuicoes["Uniforme"]["dados"]),
    "primeiro_quartil": np.percentile(distribuicoes["Uniforme"]["dados"], 25),
    "terceiro_quartil": np.percentile(distribuicoes["Uniforme"]["dados"], 75)
}

distribuicoes["Poisson"]["estatisticas"] = {
    "média": np.mean(distribuicoes["Poisson"]["dados"]),
    "mediana": np.median(distribuicoes["Poisson"]["dados"]),
    "variancia": np.var(distribuicoes["Poisson"]["dados"]),
    "desvio_padrao": np.std(distribuicoes["Poisson"]["dados"]),
    "min": np.min(distribuicoes["Poisson"]["dados"]),
    "max": np.max(distribuicoes["Poisson"]["dados"]),
    "primeiro_quartil": np.percentile(distribuicoes["Poisson"]["dados"], 25),
    "terceiro_quartil": np.percentile(distribuicoes["Poisson"]["dados"], 75)
}
data2 = {
  "Normal": distribuicoes["Normal"]["estatisticas"],
  "Exponencial": distribuicoes["Exponencial"]["estatisticas"],
  "Uniforme": distribuicoes["Uniforme"]["estatisticas"],
  "Poisson": distribuicoes["Poisson"]["estatisticas"]
}
data2 = pd.DataFrame(data2)
print(data2)
data2.T
```

```python
#fazendo todos os graficos de boxplot de uma vez
data_ = [distribuicoes["Normal"]["dados"], distribuicoes["Exponencial"]["dados"], distribuicoes["Uniforme"]["dados"], distribuicoes["Poisson"]["dados"]]
plt.boxplot(data_, positions=[1, 2, 3, 4], labels= ['Normal', 'Exponencial', 'Uniforme', 'Poisson']) 
plt.title('Gráficos de todas as distribuições juntas')
plt.show()
```
