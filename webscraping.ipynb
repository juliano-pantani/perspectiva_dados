```python
import requests
from bs4 import BeautifulSoup
import time
import pandas as pd # Usado para organizar os dados em tabela e salvar
```

```python
# URL da vítima
url = "https://myanimelist.net/anime/season"

# Cabeçalho para simular um navegador
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}
```

```python
response = requests.get(url, headers=headers)
response
```

```python
response.text
soup = BeautifulSoup(response.text, "html.parser")
soup
```

```python
# Localizando o bloco com o id "textoMateria"
bloco_noticias = soup.find("div", id="content", class_="pl0 pr0")

bloco_noticias
```

```python
b = soup.find_all("h2",class_="h2_anime_title")
todos_os_links = []
for i in range(0, len(b)):
    links = [a['href'] for a in b[i].find_all('a', href=True)]
    #todos_os_links.append(links)
    todos_os_links = todos_os_links + links
todos_os_links
todos_os_links = [link for link in todos_os_links if "https://myanimelist.net/anime/" in link]

for link in todos_os_links:
    print(link)
#links
#        
# Exibindo os links extraídos
#for link in links:
#    print(link)
```
