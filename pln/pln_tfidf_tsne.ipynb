```python
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.manifold import TSNE

df = pd.read_csv("dados_final.csv")

df.head()
```

```python
from nltk.tokenize import wordpunct_tokenize
from unidecode import unidecode


def limpar_texto(texto):
    texto = texto.lower()
    texto = unidecode(texto)
    tokens = wordpunct_tokenize(texto)
    tokens = [token for token in tokens if token.isalnum()]
    return " ".join(tokens)


df["texto_limpo"] = df["Sinopse"].apply(limpar_texto)

df[["Sinopse", "texto_limpo"]].head()
```

```python
import nltk

from nltk.corpus import stopwords


nltk.download("stopwords", quiet=True)

stopwords_en = stopwords.words("english")
stopwords_en = [unidecode(palavra) for palavra in stopwords_en]
stopwords_en = set(stopwords_en)


def remover_stopwords(texto):
    tokens = texto.split()
    tokens = [token for token in tokens if token not in stopwords_en]
    return tokens


df["tokens_sem_stopwords"] = df["texto_limpo"].apply(remover_stopwords)
df["texto_sem_stopwords"] = df["tokens_sem_stopwords"].str.join(" ")

df[["texto_limpo", "tokens_sem_stopwords", "texto_sem_stopwords"]].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
matriz_bow = vectorizer.fit_transform(df["texto_sem_stopwords"])

df_bow = pd.DataFrame(
    matriz_bow.toarray(),
    columns=vectorizer.get_feature_names_out()
)

df_bow.head()
```

```python
colunas_com_numeros = [col for col in df_bow.columns if any(char.isdigit() for char in col)]

df_bow = df_bow.drop(columns=colunas_com_numeros)

print(f"{len(colunas_com_numeros)} colunas removidas")
df_bow.head()
```

```python
metadados = df[["Nome", "Nota", "Data", "Genêros", "Material Fonte", "Tipo", "Número de episódios", "Classificação", "Posicionamento no ranking", "Posicionamento de popularidade", "Número de membros", "Favoritos", "Foto", "Sinopse"]].reset_index(drop=True)
bow_com_prefixo = df_bow.add_prefix("bow_").reset_index(drop=True)
```

```python
df_final = pd.concat([metadados, bow_com_prefixo], axis=1)

df_final.head()
```

```python
colunas_bow = [coluna for coluna in df_final.columns if coluna.startswith("bow_")]

frequencia_palavras = df_final[colunas_bow].sum().sort_values(ascending=False)

print(f"Total de palavras diferentes: {len(frequencia_palavras)}")

frequencia_palavras.head(10)
```

```python
frequencia_palavras.tail(10)
```

```python
documentos_por_palavra = (df_final[colunas_bow] > 0).sum().sort_values(ascending=False)
documentos_por_palavra.index = documentos_por_palavra.index.str.replace("bow_", "", regex=False)

documentos_por_palavra.head(10)
```

```python
df_final["palavras_unicas"] = (df_final[colunas_bow] > 0).sum(axis=1)

df_final[["Nome", "palavras_unicas"]].sort_values("palavras_unicas", ascending=False).head(10)
```

TF-IDF

```python
matriz_bow_df = df_final[colunas_bow].copy()
matriz_bow_df.columns = matriz_bow_df.columns.str.replace("bow_", "", regex=False)

matriz_bow_df.head()

total_palavras_por_noticia = matriz_bow_df.sum(axis=1)

matriz_tf = matriz_bow_df.div(total_palavras_por_noticia, axis=0)

matriz_tf.head()


noticia = 0

matriz_tf.loc[noticia].sort_values(ascending=False).head(10)

import numpy as np

N = len(matriz_bow_df)
documentos_por_termo = (matriz_bow_df > 0).sum(axis=0)

idf = np.log(N / documentos_por_termo)

df_idf = pd.DataFrame({
    "documentos_com_o_termo": documentos_por_termo,
    "idf": idf
}).sort_values("idf", ascending=False)

df_idf.head(10)


matriz_tfidf = matriz_tf.mul(idf, axis=1)

matriz_tfidf.head()

noticia = 10

matriz_tfidf.loc[noticia].sort_values(ascending=False).head(10)

tfidf_com_prefixo = matriz_tfidf.add_prefix("tfidf_").reset_index(drop=True)

df_final_tfidf = pd.concat([metadados, tfidf_com_prefixo], axis=1)

df_final_tfidf.head()

tfidf_com_prefixo.iloc[2].sort_values()

```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer_tfidf = TfidfVectorizer()
matriz_tfidf_sklearn = vectorizer_tfidf.fit_transform(df["texto_sem_stopwords"])

df_tfidf_sklearn = pd.DataFrame(
    matriz_tfidf_sklearn.toarray(),
    columns=vectorizer_tfidf.get_feature_names_out()
)

df_tfidf_sklearn.head()
```

```python
colunas_bow = [c for c in df_final.columns if c.startswith("bow_")]
colunas_tfidf = [c for c in df_final_tfidf.columns if c.startswith("tfidf_")]

X_bow = df_final[colunas_bow].values
X_tfidf = df_final_tfidf[colunas_tfidf].values

titulos = df_final["Nome"].tolist()

print(f"X_bow:   {X_bow.shape}")
print(f"X_tfidf: {X_tfidf.shape}")
```

```python
perplexity = 10
random_state = 30

tsne_bow = TSNE(n_components=2, perplexity=perplexity, early_exaggeration=5, metric="nan_euclidean", random_state=random_state, init="random")
embed_bow = tsne_bow.fit_transform(X_bow)

tsne_tfidf = TSNE(n_components=2, perplexity=perplexity, early_exaggeration=5, metric="nan_euclidean", random_state=random_state, init="random")
embed_tfidf = tsne_tfidf.fit_transform(X_tfidf)

print("embed_bow:  ", embed_bow.shape)
print("embed_tfidf:", embed_tfidf.shape)
```

```python
def plotar(embed, nome):
    df_plot = pd.DataFrame({
        "x": embed[:, 0],
        "y": embed[:, 1],
        "indice": df_final.index,
        "Nome": df_final["Nome"],
        "data": df_final["Data"],
        "generos": df_final["Genêros"]
    })
    fig = px.scatter(
        df_plot,
        x="x",
        y="y",
        hover_data={"indice": True, "Nome": True, "data": True,  "generos": True, "x": False, "y": False},
        title=f"t-SNE - {nome}",
        width=700,
        height=700,
    )
    fig.write_html(nome+".html")
    fig.show()


plotar(embed_bow, "Bag of Words")
plotar(embed_tfidf, "TF-IDF")
```
